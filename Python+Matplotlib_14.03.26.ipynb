```python
import  numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
months = ['Jan','Feb','Mar','Apr','May']
sales = [200,250,300,280,350]
```

```python
months
```

```python
sales
```

```python
plt.plot(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.plot(months, sales,marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
months = ['Jan','Feb','Mar','Apr','May']
sales = [200,250,300,280,350]
```

```python
months
```

```python
sales
```

```python
plt.figure(figsize= (8,4))
plt.plot(months, sales , marker='o', linestyle='--')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize= (8,4))
plt.scatter(months, sales)
plt.title("Monthly Salary Scatter")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize= (8,4))
plt.scatter(months, sales, s = 200, color = 'red')
plt.title("Monthly Sales Scatter")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize= (10,4))

plt.subplot(1,3,1)
plt.plot(months, sales, marker='o', color = 'red', linestyle ='--')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)

plt.subplot(1,3,2)
plt.bar(months, sales, color = 'orange')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(axis="y")

plt.subplot(1,3,3)
plt.scatter(months, sales, s = 200, color = 'green')
plt.title("Monthly Sales Scatter")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
